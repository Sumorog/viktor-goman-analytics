# A/B-тест: сравнение конверсии

Демо на синтетических числах. Группа A — контроль, B — новый тарифный экран.

```python
import math

n_a, conv_a = 1200, 180
n_b, conv_b = 1180, 212

p_a = conv_a / n_a
p_b = conv_b / n_b
diff = p_b - p_a

# SE разности долей (независимые выборки)
se = math.sqrt(p_a * (1 - p_a) / n_a + p_b * (1 - p_b) / n_b)
z = 1.96
ci_low = diff - z * se
ci_high = diff + z * se

print(f"Конверсия A: {p_a:.3%}")
print(f"Конверсия B: {p_b:.3%}")
print(f"Разница (B-A): {diff:.3%}")
print(f"95% ДИ для разницы: [{ci_low:.3%}; {ci_high:.3%}]")
if ci_low > 0:
    print("Вывод: B статистически выше A на уровне 5% (грубая проверка).")
elif ci_high < 0:
    print("Вывод: B ниже A.")
else:
    print("Вывод: эффект не установлен (ДИ пересекает 0).")
```
